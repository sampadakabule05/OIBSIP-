```python
# Unemployment Analysis – Oasis Infobyte

# Step 1: Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Step 2: Load Dataset
# Example dataset (replace with your provided unemployment dataset CSV if given)
data = {
    'State': ['Karnataka', 'Maharashtra', 'Bihar', 'Kerala', 'Punjab'],
    'Unemployment Rate': [4.5, 5.2, 7.8, 3.9, 4.8]
}
df = pd.DataFrame(data)

# If you have a CSV file instead:
# df = pd.read_csv('unemployment.csv')

# Step 3: Data Overview
print(df.head())
print("\nDataset Info:")
print(df.info())
print("\nMissing Values:")
print(df.isnull().sum())

# Step 4: Basic Statistics
print("\nDescriptive Statistics:")
print(df.describe())

# Step 5: Visualization – Bar Plot
plt.figure(figsize=(10,6))
sns.barplot(x='State', y='Unemployment Rate', data=df)
plt.title('Unemployment Rate by State')
plt.ylabel('Unemployment Rate (%)')
plt.xlabel('State')
plt.show()

# Step 6: Visualization – Pie Chart
plt.figure(figsize=(8,8))
plt.pie(df['Unemployment Rate'], labels=df['State'], autopct='%1.1f%%')
plt.title('Unemployment Rate Distribution by State')
plt.show()
```
