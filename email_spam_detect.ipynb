```python
# Email Spam Detection – Oasis Infobyte

# Step 1: Import Libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Step 2: Load Dataset
# Download spam.csv from UCI ML repo and place in same folder
df = pd.read_csv('/content/sample_data/spam.csv', encoding='latin-1')[['v1', 'v2']]
df.columns = ['label', 'message']

# Step 3: Encode Labels
df['label'] = df['label'].map({'ham': 0, 'spam': 1})

# Step 4: Split Data
X_train, X_test, y_train, y_test = train_test_split(df['message'], df['label'], test_size=0.2, random_state=42)

# Step 5: Vectorize Text
cv = CountVectorizer()
X_train_cv = cv.fit_transform(X_train)
X_test_cv = cv.transform(X_test)

# Step 6: Train Model
model = MultinomialNB()
model.fit(X_train_cv, y_train)

# Step 7: Evaluate
y_pred = model.predict(X_test_cv)
print("Accuracy:", accuracy_score(y_test, y_pred))

# Step 8: Single Email Prediction Function
def predict_email(text):
    vect = cv.transform([text])
    result = model.predict(vect)
    return 'Spam' if result[0]==1 else 'Not Spam'

# Example
print(predict_email("Congratulations! You've won a free iPhone. Call now to claim."))
```
